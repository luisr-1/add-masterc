```python

#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
ANÁLISE COMPLETA DE DESEMPENHO - Calculadora add-masterc
Análise de memória alocada, tempo de execução e eficiência computacional
Foco: Todas as operações até 10.000 dígitos + Fatorial até 500
Autor: Luis Rodrigues da Silva & Arthur Hilario Tembikoski
"""

import subprocess
import time
import json
import random
import statistics
import matplotlib.pyplot as plt
import numpy as np
import sys
import os
import psutil
from datetime import datetime
from typing import Dict, List, Tuple, Optional, Any
import warnings
import math
import re
warnings.filterwarnings('ignore', category=RuntimeWarning)

# Aumentar limite do Python para strings grandes
sys.set_int_max_str_digits(30000)

# Instalar dependências
!pip install matplotlib numpy tqdm psutil -q
from tqdm import tqdm


class PerformanceAnalyzer:
    """Analisador completo de desempenho da calculadora"""

    def __init__(self, binary_path: str = "/content/add-masterc/build/add-masterc"):
        self.binary_path = binary_path
        self.timeout = 300  # 5 minutos timeout
        self.max_digits = 10000  # Máximo de 10.000 dígitos para operações
        self.max_factorial = 500  # Máximo para fatorial
        self.operations = ['+', '-', '*', '/', '%']  # Todas as operações
        self.timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        self.results_dir = f"/content/complete_analysis_{self.timestamp}"
        os.makedirs(self.results_dir, exist_ok=True)

        self.results = {
            'operations_data': {},
            'factorial_data': [],
            'performance_summary': {},
            'limits_summary': {},
            'efficiency_data': {}
        }

    def verify_binary(self) -> bool:
        """Verifica se o binário está pronto"""
        print("="*70)
        print("ANÁLISE COMPLETA DE DESEMPENHO - add-masterc")
        print("="*70)

        if not os.path.exists(self.binary_path):
            print(f"❌ Arquivo não encontrado: {self.binary_path}")
            return False

        try:
            if not os.access(self.binary_path, os.X_OK):
                os.chmod(self.binary_path, 0o755)
        except:
            pass

        print(f"✅ Binário encontrado: {self.binary_path}")
        print(f"📊 Análises incluídas:")
        print(f"  • Memória alocada vs Número de dígitos")
        print(f"  • Tempo de execução vs Número de dígitos")
        print(f"  • Eficiência computacional vs Número de dígitos")
        print(f"  • Limites máximos testados por operação")
        print(f"  • Fatorial: memória alocada vs dígitos do resultado (até {self.max_factorial})")
        print(f"🔢 Operações: {', '.join(self.operations)}")
        print(f"⏱️  Timeout: {self.timeout}s ({self.timeout/60:.1f} minutos)")
        return True

    def get_process_memory_info(self, pid: int) -> Dict[str, float]:
        """Obtém informações de memória do processo"""
        try:
            process = psutil.Process(pid)
            mem_info = process.memory_info()
            return {
                'rss': mem_info.rss / (1024 * 1024),  # Resident Set Size (MB)
                'vms': mem_info.vms / (1024 * 1024),  # Virtual Memory Size (MB)
            }
        except (psutil.NoSuchProcess, psutil.AccessDenied):
            return {'rss': 0, 'vms': 0}

    def execute_operation(self, operation: str, num1: str, num2: Optional[str] = None) -> Dict[str, Any]:
        """
        Executa uma operação e coleta métricas de desempenho
        """
        inputs = [f"{operation}\n", f"{num1}\n"]
        if operation != '!' and num2 is not None:
            inputs.append(f"{num2}\n")
        inputs.append("q\n")

        input_str = ''.join(inputs)
        start_time = time.time()

        memory_readings = []
        timestamps = []
        peak_memory = 0
        process_pid = None

        try:
            process = subprocess.Popen(
                [self.binary_path],
                stdin=subprocess.PIPE,
                stdout=subprocess.PIPE,
                stderr=subprocess.PIPE,
                text=True,
                bufsize=1,
                universal_newlines=True
            )

            process_pid = process.pid

            # Enviar entrada
            process.stdin.write(input_str)
            process.stdin.flush()

            # Monitorar durante execução
            check_interval = 0.05  # 50ms
            max_checks = 200  # Máximo de verificações

            for check in range(max_checks):
                if process.poll() is not None:
                    break

                if time.time() - start_time > self.timeout:
                    process.terminate()
                    try:
                        process.wait(timeout=5)
                    except subprocess.TimeoutExpired:
                        process.kill()
                    raise subprocess.TimeoutExpired(cmd=self.binary_path, timeout=self.timeout)

                current_memory = self.get_process_memory_info(process_pid)
                current_rss = current_memory['rss']
                memory_readings.append(current_rss)
                timestamps.append(time.time() - start_time)

                if current_rss > peak_memory:
                    peak_memory = current_rss

                time.sleep(check_interval)

            stdout, stderr = process.communicate()
            exec_time = time.time() - start_time

            if process.returncode != 0:
                return {
                    'success': False,
                    'error': f"Exit code {process.returncode}",
                    'time': exec_time,
                    'peak_memory': peak_memory
                }

            # Extrair resultado
            output = stdout
            numbers = re.findall(r'-?\d+', output)

            if numbers:
                result_num = max(numbers, key=len)
                return {
                    'success': True,
                    'result': result_num,
                    'result_digits': len(result_num),
                    'time': exec_time,
                    'peak_memory': peak_memory,
                    'memory_readings': memory_readings,
                    'timestamps': timestamps
                }
            else:
                return {
                    'success': False,
                    'error': "Nenhum número encontrado na saída",
                    'time': exec_time,
                    'peak_memory': peak_memory
                }

        except subprocess.TimeoutExpired:
            exec_time = time.time() - start_time
            return {
                'success': False,
                'error': f"Timeout ({self.timeout}s) excedido",
                'time': exec_time,
                'peak_memory': peak_memory
            }
        except Exception as e:
            exec_time = time.time() - start_time if 'start_time' in locals() else 0
            return {
                'success': False,
                'error': str(e),
                'time': exec_time,
                'peak_memory': 0
            }
        finally:
            if process_pid:
                try:
                    process.terminate()
                    process.wait(timeout=2)
                except:
                    try:
                        process.kill()
                    except:
                        pass

    def generate_large_number(self, num_digits: int) -> str:
        """Gera número grande"""
        if num_digits <= 0:
            return "0"

        first_digit = str(random.randint(1, 9))

        if num_digits == 1:
            return first_digit

        remaining = num_digits - 1
        chunks = []

        while remaining > 0:
            chunk_size = min(1000, remaining)
            chunk = ''.join(str(random.randint(0, 9)) for _ in range(chunk_size))
            chunks.append(chunk)
            remaining -= chunk_size

        return first_digit + ''.join(chunks)

    def generate_numbers_for_operation(self, operation: str, digit_size: int) -> Tuple[str, str]:
        """Gera números apropriados para cada operação"""
        if operation == '+':
            num1 = self.generate_large_number(digit_size)
            num2 = self.generate_large_number(digit_size)

        elif operation == '-':
            num1 = self.generate_large_number(digit_size)
            num2 = self.generate_large_number(max(1, digit_size - 100))
            while len(num1) < len(num2) or (len(num1) == len(num2) and num1 < num2):
                num2 = self.generate_large_number(max(1, digit_size - 200))

        elif operation == '*':
            size1 = digit_size // 2
            size2 = digit_size - size1
            num1 = self.generate_large_number(size1)
            num2 = self.generate_large_number(size2)

        elif operation == '/':
            num1 = self.generate_large_number(digit_size)
            divisor_digits = max(1, digit_size // 3)
            num2 = self.generate_large_number(divisor_digits)
            while num2 == '0' * len(num2):
                num2 = self.generate_large_number(divisor_digits)

        elif operation == '%':
            num1 = self.generate_large_number(digit_size)
            size2 = random.randint(max(1, digit_size - 100), digit_size)
            num2 = self.generate_large_number(size2)
            while num2 == '0' * len(num2):
                num2 = self.generate_large_number(size2)

        else:
            num1 = self.generate_large_number(digit_size)
            num2 = self.generate_large_number(digit_size)

        return num1, num2

    def analyze_operations_performance(self):
        """
        Analisa desempenho de todas as operações até 10.000 dígitos
        """
        print("\n" + "="*70)
        print("ANÁLISE DE DESEMPENHO DAS OPERAÇÕES")
        print("="*70)
        print(f"🔢 Operações: {', '.join(self.operations)}")
        print(f"📊 Métricas: Memória, Tempo, Eficiência")
        print(f"⏱️  Timeout: {self.timeout}s por teste")
        print("-"*70)

        # Pontos de teste
        digit_points = [
            100, 250, 500, 750, 1000, 1500, 2000, 3000, 4000, 5000,
            6000, 7000, 8000, 9000, 10000
        ]

        operations_data = {}
        limits_summary = {}

        for operation in self.operations:
            print(f"\n🔧 Operação: {operation}")
            print("-"*50)

            operation_data = []
            max_successful_digits = 0

            for digits in tqdm(digit_points, desc=f"Testando dígitos"):
                if digits > self.max_digits:
                    continue

                # Executar 3 vezes para média
                memories = []
                times = []
                successes = 0

                for attempt in range(3):
                    try:
                        num1, num2 = self.generate_numbers_for_operation(operation, digits)
                        result = self.execute_operation(operation, num1, num2)

                        if result['success']:
                            memories.append(result['peak_memory'])
                            times.append(result['time'])
                            successes += 1

                        time.sleep(0.3)  # Pausa entre tentativas

                    except Exception as e:
                        print(f"  ⚠️ Erro em {digits} dígitos: {str(e)[:50]}")
                        continue

                if successes > 0:
                    max_successful_digits = max(max_successful_digits, digits)

                    avg_memory = statistics.mean(memories)
                    avg_time = statistics.mean(times)
                    std_memory = statistics.stdev(memories) if len(memories) > 1 else 0
                    std_time = statistics.stdev(times) if len(times) > 1 else 0

                    # Calcular eficiência computacional (dígitos por segundo)
                    computational_efficiency = digits / avg_time if avg_time > 0 else 0

                    # Calcular eficiência de memória (dígitos por MB)
                    memory_efficiency = digits / avg_memory if avg_memory > 0 else 0

                    data_point = {
                        'digits': digits,
                        'avg_memory': avg_memory,  # MB
                        'std_memory': std_memory,
                        'avg_time': avg_time,  # segundos
                        'std_time': std_time,
                        'samples': successes,
                        'computational_efficiency': computational_efficiency,  # dígitos/s
                        'memory_efficiency': memory_efficiency,  # dígitos/MB
                        'memory_per_digit': avg_memory / digits if digits > 0 else 0,
                        'time_per_digit': avg_time / digits if digits > 0 else 0
                    }

                    operation_data.append(data_point)

                    print(f"  ✓ {digits:5,d} dígitos: "
                          f"Mem: {avg_memory:6.1f} MB | "
                          f"Tempo: {avg_time:5.2f}s | "
                          f"Efic: {computational_efficiency:6.0f} dígitos/s")

            operations_data[operation] = operation_data
            limits_summary[operation] = max_successful_digits

            print(f"\n  🏆 Limite alcançado: {max_successful_digits:,} dígitos")

            # Calcular estatísticas da operação
            if operation_data:
                memories = [d['avg_memory'] for d in operation_data]
                times = [d['avg_time'] for d in operation_data]
                efficiencies = [d['computational_efficiency'] for d in operation_data]

                stats = {
                    'max_memory': max(memories),
                    'avg_memory': statistics.mean(memories),
                    'max_time': max(times),
                    'avg_time': statistics.mean(times),
                    'max_efficiency': max(efficiencies),
                    'avg_efficiency': statistics.mean(efficiencies)
                }

                self.results['performance_summary'][operation] = stats

        self.results['operations_data'] = operations_data
        self.results['limits_summary'] = limits_summary

        # Salvar dados
        with open(f"{self.results_dir}/operations_performance.json", 'w') as f:
            json.dump(operations_data, f, indent=2)

        print(f"\n✅ Dados das operações salvos em: {self.results_dir}/operations_performance.json")

    def analyze_factorial_memory_allocation(self):
        """
        Analisa memória alocada para fatorial vs dígitos do resultado
        """
        print("\n" + "="*70)
        print("ANÁLISE DE FATORIAL: MEMÓRIA vs DÍGITOS DO RESULTADO")
        print("="*70)
        print(f"🔢 Fatorial até: {self.max_factorial}")
        print(f"📊 Análise: Memória alocada vs dígitos do resultado")
        print(f"⏱️  Timeout: {self.timeout}s por teste")
        print("-"*70)

        # Pontos de teste para fatorial
        factorial_points = [
            1, 5, 10, 20, 30, 40, 50, 75, 100, 125,
            150, 175, 200, 225, 250, 275, 300, 325,
            350, 375, 400, 425, 450, 475, 500
        ]

        factorial_data = []

        for n in tqdm(factorial_points, desc="Testando fatoriais"):
            if n > self.max_factorial:
                continue

            # Executar 3 vezes para média
            memories = []
            times = []
            result_digits_list = []
            successes = 0

            for attempt in range(3):
                try:
                    result = self.execute_operation('!', str(n), None)

                    if result['success']:
                        memories.append(result['peak_memory'])
                        times.append(result['time'])
                        result_digits_list.append(result['result_digits'])
                        successes += 1

                    time.sleep(0.5)  # Pausa entre tentativas

                except Exception as e:
                    print(f"  ⚠️ Erro em {n}!: {str(e)[:50]}")
                    continue

            if successes > 0:
                avg_memory = statistics.mean(memories)
                avg_time = statistics.mean(times)
                avg_result_digits = statistics.mean(result_digits_list)

                # Calcular dígitos teóricos (fórmula de Stirling)
                if n > 0:
                    expected_digits = int(n * math.log10(n) - n * math.log10(math.e) +
                                        0.5 * math.log10(2 * math.pi * n)) + 1
                else:
                    expected_digits = 1

                # Calcular eficiências
                memory_per_digit = avg_memory / avg_result_digits if avg_result_digits > 0 else 0
                computational_efficiency = avg_result_digits / avg_time if avg_time > 0 else 0

                data_point = {
                    'n': n,
                    'avg_memory': avg_memory,  # MB
                    'avg_time': avg_time,  # segundos
                    'result_digits': avg_result_digits,
                    'expected_digits': expected_digits,
                    'digit_error': abs(avg_result_digits - expected_digits),
                    'memory_per_digit': memory_per_digit,
                    'computational_efficiency': computational_efficiency,  # dígitos/s
                    'samples': successes
                }

                factorial_data.append(data_point)

                print(f"  ✓ {n:3}! → {avg_result_digits:6,.0f} dígitos | "
                      f"Mem: {avg_memory:6.1f} MB | "
                      f"Tempo: {avg_time:6.2f}s | "
                      f"Mem/dígito: {memory_per_digit:.4f} MB")

        self.results['factorial_data'] = factorial_data

        # Salvar dados
        with open(f"{self.results_dir}/factorial_analysis.json", 'w') as f:
            json.dump(factorial_data, f, indent=2)

        print(f"\n✅ Dados do fatorial salvos em: {self.results_dir}/factorial_analysis.json")

    def create_consolidated_charts(self):
        """Cria todos os gráficos consolidados"""
        print("\n" + "="*70)
        print("GERANDO GRÁFICOS CONSOLIDADOS")
        print("="*70)

        # Configurações de cores
        colors = {
            '+': '#4CAF50',  # Verde
            '-': '#2196F3',  # Azul
            '*': '#FF9800',  # Laranja
            '/': '#9C27B0',  # Roxo
            '%': '#F44336'   # Vermelho
        }

        op_names = {
            '+': 'Adição',
            '-': 'Subtração',
            '*': 'Multiplicação',
            '/': 'Divisão',
            '%': 'Módulo'
        }

        # 1. GRÁFICO: Memória Alocada vs Número de Dígitos
        print("📊 Gerando gráfico: Memória Alocada vs Número de Dígitos")
        self._create_memory_vs_digits_chart(colors, op_names)

        # 2. GRÁFICO: Tempo de Execução vs Número de Dígitos
        print("📊 Gerando gráfico: Tempo de Execução vs Número de Dígitos")
        self._create_time_vs_digits_chart(colors, op_names)

        # 3. GRÁFICO: Limites de Dígitos Testados por Operação
        print("📊 Gerando gráfico: Limites de Dígitos Testados")
        self._create_limits_chart(colors, op_names)

        # 4. GRÁFICO: Eficiência Computacional vs Número de Dígitos
        print("📊 Gerando gráfico: Eficiência Computacional")
        self._create_efficiency_vs_digits_chart(colors, op_names)

        # 5. GRÁFICO: Fatorial - Memória vs Dígitos do Resultado
        print("📊 Gerando gráfico: Fatorial - Memória vs Dígitos")
        self._create_factorial_memory_chart()

        # 6. GRÁFICO: Visão Consolidada (4 em 1)
        print("📊 Gerando gráfico: Visão Consolidada")
        self._create_consolidated_view_chart(colors, op_names)

        print(f"\n✅ Todos os gráficos salvos em: {self.results_dir}/")

    def _create_memory_vs_digits_chart(self, colors: Dict, op_names: Dict):
        """Gráfico: Memória Alocada vs Número de Dígitos"""
        if not self.results.get('operations_data'):
            return

        plt.figure(figsize=(14, 8))

        for operation in self.operations:
            data = self.results['operations_data'].get(operation, [])
            if data:
                digits = [d['digits'] for d in data]
                memories = [d['avg_memory'] for d in data]

                plt.plot(digits, memories, 'o-',
                        color=colors.get(operation, '#000000'),
                        linewidth=3,
                        markersize=8,
                        markerfacecolor='white',
                        markeredgewidth=2,
                        label=f"{op_names.get(operation, operation)}")

        plt.xlabel('Número de Dígitos', fontsize=13, fontweight='bold')
        plt.ylabel('Memória Alocada (MB)', fontsize=13, fontweight='bold')
        plt.title('MEMÓRIA ALOCADA vs NÚMERO DE DÍGITOS\n' +
                 f'Operações aritméticas | Timeout: {self.timeout}s',
                 fontsize=15, fontweight='bold', pad=20)

        plt.grid(True, alpha=0.3, linestyle='--')
        plt.legend(fontsize=11, loc='upper left')
        plt.xlim(0, self.max_digits * 1.05)

        # Adicionar linhas de referência
        for mb_line in [100, 500, 1000, 2000, 5000]:
            plt.axhline(y=mb_line, color='gray', linestyle=':', alpha=0.3)

        plt.tight_layout()
        plt.savefig(f"{self.results_dir}/memory_vs_digits.png", dpi=150, bbox_inches='tight')
        plt.close()

    def _create_time_vs_digits_chart(self, colors: Dict, op_names: Dict):
        """Gráfico: Tempo de Execução vs Número de Dígitos"""
        if not self.results.get('operations_data'):
            return

        plt.figure(figsize=(14, 8))

        for operation in self.operations:
            data = self.results['operations_data'].get(operation, [])
            if data:
                digits = [d['digits'] for d in data]
                times = [d['avg_time'] for d in data]

                plt.plot(digits, times, 's-',
                        color=colors.get(operation, '#000000'),
                        linewidth=3,
                        markersize=8,
                        markerfacecolor='white',
                        markeredgewidth=2,
                        label=f"{op_names.get(operation, operation)}")

        # Linha de timeout
        plt.axhline(y=self.timeout, color='red', linestyle='--', alpha=0.7, linewidth=2)
        plt.text(self.max_digits * 0.7, self.timeout * 1.05,
                f'Timeout: {self.timeout}s', fontsize=10, color='red', fontweight='bold')

        plt.xlabel('Número de Dígitos', fontsize=13, fontweight='bold')
        plt.ylabel('Tempo de Execução (s)', fontsize=13, fontweight='bold')
        plt.title('TEMPO DE EXECUÇÃO vs NÚMERO DE DÍGITOS\n' +
                 f'Operações aritméticas | Timeout: {self.timeout}s',
                 fontsize=15, fontweight='bold', pad=20)

        plt.grid(True, alpha=0.3, linestyle='--')
        plt.legend(fontsize=11, loc='upper left')
        plt.xlim(0, self.max_digits * 1.05)
        plt.ylim(0, self.timeout * 1.2)

        plt.tight_layout()
        plt.savefig(f"{self.results_dir}/time_vs_digits.png", dpi=150, bbox_inches='tight')
        plt.close()

    def _create_limits_chart(self, colors: Dict, op_names: Dict):
        """Gráfico: Limites de Dígitos Testados por Operação"""
        if not self.results.get('limits_summary'):
            return

        plt.figure(figsize=(12, 7))

        operations = list(self.results['limits_summary'].keys())
        limits = list(self.results['limits_summary'].values())

        bar_colors = [colors.get(op, '#607D8B') for op in operations]
        op_labels = [op_names.get(op, op) for op in operations]

        bars = plt.bar(op_labels, limits, color=bar_colors, edgecolor='black', linewidth=2)

        plt.xlabel('Operação', fontsize=12, fontweight='bold')
        plt.ylabel('Máximo de Dígitos Testados', fontsize=12, fontweight='bold')
        plt.title('LIMITES MÁXIMOS TESTADOS POR OPERAÇÃO\n' +
                 f'Timeout: {self.timeout}s | Máximo: {self.max_digits:,} dígitos',
                 fontsize=14, fontweight='bold', pad=20)

        # Adicionar valores nas barras
        for bar, limit in zip(bars, limits):
            height = bar.get_height()
            plt.text(bar.get_x() + bar.get_width()/2., height + 100,
                    f'{limit:,}', ha='center', va='bottom',
                    fontsize=11, fontweight='bold', color='darkblue')

        # Linha do máximo teórico
        plt.axhline(y=self.max_digits, color='red', linestyle='--', alpha=0.5, linewidth=2)
        plt.text(len(operations)-0.5, self.max_digits * 1.02,
                f'Máximo testado: {self.max_digits:,} dígitos',
                fontsize=10, color='red', fontweight='bold')

        plt.grid(True, alpha=0.3, axis='y', linestyle='--')
        plt.ylim(0, self.max_digits * 1.3)

        plt.tight_layout()
        plt.savefig(f"{self.results_dir}/limits_chart.png", dpi=150, bbox_inches='tight')
        plt.close()

    def _create_efficiency_vs_digits_chart(self, colors: Dict, op_names: Dict):
        """Gráfico: Eficiência Computacional vs Número de Dígitos"""
        if not self.results.get('operations_data'):
            return

        plt.figure(figsize=(14, 8))

        for operation in self.operations:
            data = self.results['operations_data'].get(operation, [])
            if data:
                digits = [d['digits'] for d in data]
                efficiencies = [d['computational_efficiency'] for d in data]

                plt.plot(digits, efficiencies, '^-',
                        color=colors.get(operation, '#000000'),
                        linewidth=2,
                        markersize=6,
                        label=f"{op_names.get(operation, operation)}")

        plt.xlabel('Número de Dígitos', fontsize=13, fontweight='bold')
        plt.ylabel('Eficiência Computacional (dígitos/s)', fontsize=13, fontweight='bold')
        plt.title('EFICIÊNCIA COMPUTACIONAL vs NÚMERO DE DÍGITOS\n' +
                 'Maior valor = maior eficiência',
                 fontsize=15, fontweight='bold', pad=20)

        plt.grid(True, alpha=0.3, linestyle='--')
        plt.legend(fontsize=11)
        plt.xlim(0, self.max_digits * 1.05)

        # Escala logarítmica para melhor visualização
        plt.yscale('log')

        plt.tight_layout()
        plt.savefig(f"{self.results_dir}/efficiency_vs_digits.png", dpi=150, bbox_inches='tight')
        plt.close()

    def _create_factorial_memory_chart(self):
        """Gráfico: Fatorial - Memória vs Dígitos do Resultado"""
        if not self.results.get('factorial_data'):
            return

        data = self.results['factorial_data']

        # Gráfico 1: Memória vs Dígitos do Resultado
        plt.figure(figsize=(14, 10))

        plt.subplot(2, 2, 1)
        result_digits = [d['result_digits'] for d in data]
        memories = [d['avg_memory'] for d in data]
        n_values = [d['n'] for d in data]

        scatter = plt.scatter(result_digits, memories, c=n_values,
                            cmap='viridis', s=80, alpha=0.7, edgecolors='black')
        plt.colorbar(scatter, label='n (fatorial)')

        plt.xlabel('Dígitos do Resultado', fontsize=11, fontweight='bold')
        plt.ylabel('Memória Alocada (MB)', fontsize=11, fontweight='bold')
        plt.title('FATORIAL: Memória Alocada vs Dígitos do Resultado',
                 fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3, linestyle='--')

        # Linha de tendência
        if len(result_digits) > 1:
            z = np.polyfit(result_digits, memories, 2)
            p = np.poly1d(z)
            digits_smooth = np.linspace(min(result_digits), max(result_digits), 100)
            plt.plot(digits_smooth, p(digits_smooth), 'r-', alpha=0.5,
                    linewidth=2, label='Tendência')
            plt.legend()

        # Gráfico 2: Memória por Dígito vs n
        plt.subplot(2, 2, 2)
        n_vals = [d['n'] for d in data]
        memory_per_digit = [d['memory_per_digit'] for d in data]

        plt.plot(n_vals, memory_per_digit, 'bo-', linewidth=2, markersize=6)

        plt.xlabel('n (fatorial)', fontsize=11, fontweight='bold')
        plt.ylabel('Memória por Dígito (MB/dígito)', fontsize=11, fontweight='bold')
        plt.title('FATORIAL: Memória por Dígito do Resultado',
                 fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3, linestyle='--')

        # Gráfico 3: Crescimento de Dígitos
        plt.subplot(2, 2, 3)
        plt.plot(n_vals, result_digits, 'go-', linewidth=2, markersize=6, label='Medido')

        # Linha teórica (Stirling)
        n_smooth = np.linspace(min(n_vals), max(n_vals), 100)
        digits_theoretical = [int(n * math.log10(n) - n * math.log10(math.e) +
                                0.5 * math.log10(2 * math.pi * n)) + 1
                            for n in n_smooth]
        plt.plot(n_smooth, digits_theoretical, 'k--', alpha=0.7,
                linewidth=2, label='Teórico (Stirling)')

        plt.xlabel('n (fatorial)', fontsize=11, fontweight='bold')
        plt.ylabel('Dígitos do Resultado', fontsize=11, fontweight='bold')
        plt.title('CRESCIMENTO DE DÍGITOS DO FATORIAL',
                 fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3, linestyle='--')
        plt.legend()

        # Gráfico 4: Eficiência Computacional
        plt.subplot(2, 2, 4)
        computational_efficiency = [d['computational_efficiency'] for d in data]

        plt.plot(n_vals, computational_efficiency, 'ro-', linewidth=2, markersize=6)

        plt.xlabel('n (fatorial)', fontsize=11, fontweight='bold')
        plt.ylabel('Eficiência (dígitos/s)', fontsize=11, fontweight='bold')
        plt.title('FATORIAL: Eficiência Computacional',
                 fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3, linestyle='--')
        plt.yscale('log')

        plt.suptitle(f'ANÁLISE COMPLETA DO FATORIAL (até {self.max_factorial})\n' +
                    f'Memória Alocada vs Dígitos do Resultado | Timeout: {self.timeout}s',
                    fontsize=16, fontweight='bold', y=0.98)
        plt.tight_layout(rect=[0, 0.03, 1, 0.95])
        plt.savefig(f"{self.results_dir}/factorial_memory_analysis.png", dpi=150, bbox_inches='tight')
        plt.close()

    def _create_consolidated_view_chart(self, colors: Dict, op_names: Dict):
        """Gráfico: Visão Consolidada (4 em 1)"""
        if not self.results.get('operations_data'):
            return

        plt.figure(figsize=(16, 12))

        # 1. Memória vs Dígitos
        plt.subplot(2, 2, 1)

        for operation in self.operations:
            data = self.results['operations_data'].get(operation, [])
            if data:
                digits = [d['digits'] for d in data]
                memories = [d['avg_memory'] for d in data]
                plt.plot(digits, memories, '-',
                        color=colors.get(operation, '#000000'),
                        linewidth=2,
                        label=op_names.get(operation, operation))

        plt.xlabel('Número de Dígitos', fontsize=11)
        plt.ylabel('Memória Alocada (MB)', fontsize=11)
        plt.title('Memória vs Dígitos', fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=9)

        # 2. Tempo vs Dígitos
        plt.subplot(2, 2, 2)

        for operation in self.operations:
            data = self.results['operations_data'].get(operation, [])
            if data:
                digits = [d['digits'] for d in data]
                times = [d['avg_time'] for d in data]
                plt.plot(digits, times, '-',
                        color=colors.get(operation, '#000000'),
                        linewidth=2,
                        label=op_names.get(operation, operation))

        plt.axhline(y=self.timeout, color='red', linestyle='--', alpha=0.5, linewidth=1)
        plt.xlabel('Número de Dígitos', fontsize=11)
        plt.ylabel('Tempo (s)', fontsize=11)
        plt.title(f'Tempo vs Dígitos (Timeout: {self.timeout}s)', fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=9)

        # 3. Eficiência vs Dígitos
        plt.subplot(2, 2, 3)

        for operation in self.operations:
            data = self.results['operations_data'].get(operation, [])
            if data:
                digits = [d['digits'] for d in data]
                efficiencies = [d['computational_efficiency'] for d in data]
                plt.plot(digits, efficiencies, '-',
                        color=colors.get(operation, '#000000'),
                        linewidth=2,
                        label=op_names.get(operation, operation))

        plt.xlabel('Número de Dígitos', fontsize=11)
        plt.ylabel('Eficiência (dígitos/s)', fontsize=11)
        plt.title('Eficiência Computacional vs Dígitos', fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=9)
        plt.yscale('log')

        # 4. Memória por Dígito vs Dígitos
        plt.subplot(2, 2, 4)

        for operation in self.operations:
            data = self.results['operations_data'].get(operation, [])
            if data:
                digits = [d['digits'] for d in data]
                memory_per_digit = [d['memory_per_digit'] for d in data]
                plt.plot(digits, memory_per_digit, '-',
                        color=colors.get(operation, '#000000'),
                        linewidth=2,
                        label=op_names.get(operation, operation))

        plt.xlabel('Número de Dígitos', fontsize=11)
        plt.ylabel('Memória por Dígito (MB/dígito)', fontsize=11)
        plt.title('Eficiência de Memória vs Dígitos', fontsize=12, fontweight='bold')
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=9)

        plt.suptitle(f'VISÃO CONSOLIDADA - ANÁLISE DE DESEMPENHO\n' +
                    f'Operações: {", ".join(self.operations)} | Dígitos máx: {self.max_digits:,} | Timeout: {self.timeout}s',
                    fontsize=16, fontweight='bold', y=0.98)
        plt.tight_layout(rect=[0, 0.03, 1, 0.95])
        plt.savefig(f"{self.results_dir}/consolidated_view.png", dpi=150, bbox_inches='tight')
        plt.close()

    def generate_comprehensive_report(self):
        """Gera relatório completo da análise"""
        print("\n" + "="*70)
        print("GERANDO RELATÓRIO COMPLETO")
        print("="*70)

        report_path = f"{self.results_dir}/comprehensive_report.txt"

        with open(report_path, 'w', encoding='utf-8') as f:
            f.write("="*80 + "\n")
            f.write("RELATÓRIO COMPLETO DE ANÁLISE DE DESEMPENHO - add-masterc\n")
            f.write("="*80 + "\n\n")

            f.write(f"Data da análise: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
            f.write(f"Binário testado: {self.binary_path}\n")
            f.write(f"Timeout por teste: {self.timeout} segundos ({self.timeout/60:.1f} minutos)\n")
            f.write(f"Dígitos máximos testados (operações): {self.max_digits:,}\n")
            f.write(f"Fatorial máximo testado: {self.max_factorial}\n")
            f.write(f"Operações analisadas: {', '.join(self.operations)}\n\n")

            # Resumo dos limites
            f.write("LIMITES MÁXIMOS TESTADOS POR OPERAÇÃO:\n")
            f.write("-"*50 + "\n")
            if self.results.get('limits_summary'):
                for op, limit in self.results['limits_summary'].items():
                    op_name = {'+': 'Adição', '-': 'Subtração', '*': 'Multiplicação',
                              '/': 'Divisão', '%': 'Módulo'}.get(op, op)
                    percentage = (limit / self.max_digits) * 100
                    f.write(f"{op_name:12} ({op}): {limit:6,} dígitos ({percentage:5.1f}% do máximo)\n")
            f.write("\n")

            # Estatísticas de desempenho
            f.write("ESTATÍSTICAS DE DESEMPENHO POR OPERAÇÃO:\n")
            f.write("-"*50 + "\n")
            if self.results.get('performance_summary'):
                f.write(f"{'Operação':12} {'Memória (MB)':>15} {'Tempo (s)':>12} {'Eficiência':>12}\n")
                f.write(f"{'':12} {'Máx':>7} {'Méd':>7} {'Máx':>6} {'Méd':>6} {'Máx':>6} {'Méd':>6}\n")
                f.write("-"*70 + "\n")

                for op, stats in self.results['performance_summary'].items():
                    op_name = {'+': 'Adição', '-': 'Subtração', '*': 'Multiplicação',
                              '/': 'Divisão', '%': 'Módulo'}.get(op, op)
                    f.write(f"{op_name:12} {stats['max_memory']:7.0f} {stats['avg_memory']:7.0f} "
                           f"{stats['max_time']:6.1f} {stats['avg_time']:6.1f} "
                           f"{stats['max_efficiency']:6,.0f} {stats['avg_efficiency']:6,.0f}\n")
            f.write("\n")

            # Análise do fatorial
            f.write("ANÁLISE DO FATORIAL:\n")
            f.write("-"*40 + "\n")
            if self.results.get('factorial_data'):
                data = self.results['factorial_data']
                if data:
                    max_n = max(d['n'] for d in data)
                    max_digits = max(d['result_digits'] for d in data)
                    max_memory = max(d['avg_memory'] for d in data)
                    avg_memory_per_digit = statistics.mean([d['memory_per_digit'] for d in data])

                    f.write(f"Maior n testado: {max_n}\n")
                    f.write(f"Maior resultado: {max_digits:,} dígitos\n")
                    f.write(f"Maior memória alocada: {max_memory:.0f} MB\n")
                    f.write(f"Média de memória por dígito: {avg_memory_per_digit:.4f} MB/dígito\n")

                    # Encontrar ponto mais eficiente
                    efficiencies = [(d['n'], d['computational_efficiency']) for d in data]
                    most_efficient = max(efficiencies, key=lambda x: x[1])
                    f.write(f"Ponto mais eficiente: {most_efficient[0]}! com {most_efficient[1]:,.0f} dígitos/s\n")
            f.write("\n")

            # Conclusões
            f.write("CONCLUSÕES E RECOMENDAÇÕES:\n")
            f.write("="*80 + "\n\n")

            if self.results.get('performance_summary') and self.results.get('limits_summary'):
                # Operação mais eficiente em memória
                memory_efficiencies = {}
                for op in self.operations:
                    data = self.results['operations_data'].get(op, [])
                    if data:
                        avg_memory_per_digit = statistics.mean([d['memory_per_digit'] for d in data])
                        memory_efficiencies[op] = avg_memory_per_digit

                if memory_efficiencies:
                    most_efficient_mem = min(memory_efficiencies.items(), key=lambda x: x[1])
                    least_efficient_mem = max(memory_efficiencies.items(), key=lambda x: x[1])

                    eff_op = {'+': 'Adição', '-': 'Subtração', '*': 'Multiplicação',
                             '/': 'Divisão', '%': 'Módulo'}.get(most_efficient_mem[0], most_efficient_mem[0])
                    ineff_op = {'+': 'Adição', '-': 'Subtração', '*': 'Multiplicação',
                               '/': 'Divisão', '%': 'Módulo'}.get(least_efficient_mem[0], least_efficient_mem[0])

                    f.write("1. EFICIÊNCIA DE MEMÓRIA:\n")
                    f.write(f"   • Mais eficiente: {eff_op} ({most_efficient_mem[1]:.4f} MB/dígito)\n")
                    f.write(f"   • Menos eficiente: {ineff_op} ({least_efficient_mem[1]:.4f} MB/dígito)\n")
                    f.write(f"   • Diferença: {least_efficient_mem[1]/most_efficient_mem[1]:.1f}x\n\n")

                # Operação mais rápida
                computational_efficiencies = {}
                for op in self.operations:
                    data = self.results['operations_data'].get(op, [])
                    if data:
                        avg_efficiency = statistics.mean([d['computational_efficiency'] for d in data])
                        computational_efficiencies[op] = avg_efficiency

                if computational_efficiencies:
                    fastest = max(computational_efficiencies.items(), key=lambda x: x[1])
                    slowest = min(computational_efficiencies.items(), key=lambda x: x[1])

                    fast_op = {'+': 'Adição', '-': 'Subtração', '*': 'Multiplicação',
                              '/': 'Divisão', '%': 'Módulo'}.get(fastest[0], fastest[0])
                    slow_op = {'+': 'Adição', '-': 'Subtração', '*': 'Multiplicação',
                              '/': 'Divisão', '%': 'Módulo'}.get(slowest[0], slowest[0])

                    f.write("2. VELOCIDADE DE PROCESSAMENTO:\n")
                    f.write(f"   • Mais rápida: {fast_op} ({fastest[1]:,.0f} dígitos/s)\n")
                    f.write(f"   • Mais lenta: {slow_op} ({slowest[1]:,.0f} dígitos/s)\n")
                    f.write(f"   • Diferença: {fastest[1]/slowest[1]:.1f}x\n\n")

                f.write("3. RECOMENDAÇÕES PARA USO PRÁTICO:\n")
                f.write("   • Para números grandes (>5.000 dígitos): Priorize operações com melhor escalabilidade\n")
                f.write("   • Monitore uso de memória em tempo real para operações complexas\n")
                f.write("   • Considere dividir operações muito grandes em partes menores\n")
                f.write("   • Use cache para resultados de fatorial frequentemente calculados\n")
                f.write("   • Configure timeout apropriado baseado no tamanho dos números\n")

            f.write("\n" + "="*80 + "\n")
            f.write("GRÁFICOS GERADOS:\n")
            f.write("="*80 + "\n")
            f.write("1. memory_vs_digits.png - Memória Alocada vs Número de Dígitos\n")
            f.write("2. time_vs_digits.png - Tempo de Execução vs Número de Dígitos\n")
            f.write("3. limits_chart.png - Limites de Dígitos Testados por Operação\n")
            f.write("4. efficiency_vs_digits.png - Eficiência Computacional vs Dígitos\n")
            f.write("5. factorial_memory_analysis.png - Análise do Fatorial\n")
            f.write("6. consolidated_view.png - Visão Consolidada (4 em 1)\n")
            f.write(f"\nTodos os arquivos disponíveis em: {self.results_dir}/\n")

        print(f"📄 Relatório salvo: {report_path}")

    def run_complete_analysis(self):
        """Executa análise completa"""
        print("="*80)
        print("ANÁLISE COMPLETA DE DESEMPENHO - CALCULADORA add-masterc")
        print("="*80)
        print(f"Início: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"Binário: {self.binary_path}")
        print(f"🔢 Operações: {', '.join(self.operations)}")
        print(f"📊 Dígitos máximos: {self.max_digits:,} (operações)")
        print(f"🧮 Fatorial máximo: {self.max_factorial}")
        print(f"⏱️  Timeout: {self.timeout}s ({self.timeout/60:.1f}min)")
        print("="*80)

        start_time = time.time()

        try:
            # 1. Verificar binário
            print("\n[1/5] 🔍 Verificando binário...")
            if not self.verify_binary():
                return

            # 2. Análise das operações
            print("\n[2/5] 📊 Analisando desempenho das operações...")
            self.analyze_operations_performance()

            # 3. Análise do fatorial
            print("\n[3/5] 🧮 Analisando fatorial...")
            self.analyze_factorial_memory_allocation()

            # 4. Gerar gráficos
            print("\n[4/5] 📈 Gerando gráficos consolidados...")
            self.create_consolidated_charts()

            # 5. Relatório
            print("\n[5/5] 📄 Gerando relatório completo...")
            self.generate_comprehensive_report()

            # Finalização
            total_time = time.time() - start_time
            print("\n" + "="*80)
            print("✅ ANÁLISE CONCLUÍDA COM SUCESSO!")
            print("="*80)

            print(f"\n📁 RESULTADOS EM: {self.results_dir}/")
            print(f"⏱️  Tempo total: {total_time/60:.1f} minutos")

            if self.results.get('limits_summary'):
                print("\n📊 RESUMO DOS LIMITES:")
                print("-"*50)
                for op, limit in self.results['limits_summary'].items():
                    op_name = {'+': 'Adição', '-': 'Subtração', '*': 'Multiplicação',
                              '/': 'Divisão', '%': 'Módulo'}.get(op, op)
                    print(f"  {op_name:12}: {limit:6,} dígitos")

            print(f"\n📈 Gráficos gerados:")
            print(f"  • Memória Alocada vs Dígitos")
            print(f"  • Tempo de Execução vs Dígitos")
            print(f"  • Limites por Operação")
            print(f"  • Eficiência Computacional vs Dígitos")
            print(f"  • Fatorial: Memória vs Dígitos do Resultado")
            print(f"  • Visão Consolidada (4 em 1)")

        except KeyboardInterrupt:
            print("\n\n❌ Análise interrompida pelo usuário")
        except Exception as e:
            print(f"\n❌ ERRO: {e}")
            import traceback
            traceback.print_exc()


def main():
    """Função principal"""
    print("🚀 ANÁLISE COMPLETA DE DESEMPENHO - add-masterc")
    print("="*70)
    print("📊 Análises incluídas:")
    print("  • Memória alocada vs Número de dígitos")
    print("  • Tempo de execução vs Número de dígitos")
    print("  • Eficiência computacional vs Número de dígitos")
    print("  • Limites máximos testados por operação")
    print("  • Fatorial: memória vs dígitos do resultado (até 500)")
    print("="*70)

    # Verificar Colab
    try:
        from IPython import get_ipython
        in_colab = 'google.colab' in str(get_ipython())
    except:
        in_colab = False

    if in_colab:
        print("✅ Executando no Google Colab")

        os.makedirs("/content/add-masterc/build", exist_ok=True)
        binary_path = "/content/add-masterc/build/add-masterc"

        print(f"\n📁 Caminho do binário: {binary_path}")

        if not os.path.exists(binary_path):
            print("❌ Arquivo não encontrado!")
            print("\n📋 INSTRUÇÕES:")
            print("1. Faça upload do binário 'add-masterc' para:")
            print("   /content/add-masterc/build/")
            print("2. Dê permissão de execução:")
            print("   !chmod +x /content/add-masterc/build/add-masterc")
            print("3. Execute este script novamente")
            return

        analyzer = PerformanceAnalyzer(binary_path=binary_path)
        analyzer.run_complete_analysis()

    else:
        print("⚠️  Executando localmente")

        import argparse
        parser = argparse.ArgumentParser(
            description='Análise completa de desempenho da calculadora'
        )
        parser.add_argument('--binary', type=str, default='./add-masterc',
                          help='Caminho para o executável da calculadora')

        args = parser.parse_args()

        analyzer = PerformanceAnalyzer(binary_path=args.binary)
        analyzer.run_complete_analysis()


if __name__ == "__main__":
    main()
```
